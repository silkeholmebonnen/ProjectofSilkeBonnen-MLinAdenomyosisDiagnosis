```python
import pandas as pd

ieee = pd.read_csv("../literature/uterus/ieee.csv")
pubmed = pd.read_csv("../literature/uterus/pubmed.csv")
lens = pd.read_csv("../literature/uterus/lens.csv")

print(ieee.shape)
print(pubmed.shape)
print(lens.shape)
```

```python
for dataset in [ieee, lens, pubmed]:
    if "Document Title" in dataset.columns:
        dataset.rename(columns={"Document Title": "Title"}, inplace=True)
    if "Authors" in dataset.columns:
        dataset.rename(columns={"Authors": "Author/s"}, inplace=True)

df = pd.concat([ieee, lens, pubmed], ignore_index=True)
df.sort_values(by="Title", ascending=True, inplace=True)

print(df.columns)
print(df.shape)

initial_number_of_papers = df.shape[0]
```

```python
df["Title"] = df["Title"].str.lower()
df["Abstract"] = df["Abstract"].str.lower()
```

```python
df.sort_values(by="Abstract", ascending=True, inplace=True)
df = df.drop_duplicates(subset="DOI", keep="first")

print(df.shape)

after_removing_duplicates = df.shape[0]

print("number of duplicate removed: ", initial_number_of_papers - df.shape[0])
```

```python
df = df[
    df["Abstract"].str.contains("transvaginal ultrasound", na=False)
    | df["Title"].str.contains("transvaginal ultrasound", na=False) | df["Abstract"].str.contains("TVUS", na=False) | df["Title"].str.contains("TVUS", na=False)
]

print(df.shape)
after_filtering_for_adenomyosis = df.shape[0]
print("number of papers removed by filtering for ultrasound: ", after_removing_duplicates - after_filtering_for_adenomyosis)
```

```python
df = df[
    df["Abstract"].str.contains("deep learning", na=False) | df["Abstract"].str.contains("machine learning", na=False) | df["Abstract"].str.contains("artificial intelligence", na=False) | df["Abstract"].str.contains("neural networks", na=False) |
    df["Title"].str.contains("deep learning", na=False) | df["Title"].str.contains("machine learning", na=False) | df["Title"].str.contains("artificial intelligence", na=False) | df["Title"].str.contains("neural networks", na=False)
]

print(df.shape)
print("number of papers removed by filtering for machine learning : ", after_filtering_for_adenomyosis - df.shape[0])
```

```python
df.sort_values(by="Title", ascending=True, inplace=True)
df.to_csv("../filtered_papers/uterus/filtered-papers.csv", index=False, columns=["Title","Author/s","DOI", "External URL"])
```
