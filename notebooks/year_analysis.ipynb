```python
import pandas as pd

adenomyosis = pd.read_csv("../filtered_papers/adenomyosis/filtered-papers_manual_post_filtering.csv")
uterus = pd.read_csv("../filtered_papers/uterus/filtered-papers_manual_post_filtering.csv")
```

```python
def analyze_publication_years(df):
    mean_year = df["Publication Year"].mean()
    median_year = df["Publication Year"].median()
    std_dev_year = df["Publication Year"].std()
    min_year = df["Publication Year"].min()
    max_year = df["Publication Year"].max()

    return mean_year, median_year, std_dev_year, min_year, max_year
```

```python
mean_year_adenomyosis, median_year_adenomyosis, std_dev_year_adenomyosis, min_year_adenomyosis, max_year_adenomyosis = analyze_publication_years(adenomyosis)
mean_year_uterus, median_year_uterus, std_dev_year_uterus, min_year_uterus, max_year_uterus = analyze_publication_years(uterus)
```

```python
print(f"Adenomyosis papers - Mean: {mean_year_adenomyosis:.2f}, Median: {median_year_adenomyosis}, Std Dev: {std_dev_year_adenomyosis:.2f}, Min: {min_year_adenomyosis}, Max: {max_year_adenomyosis}")

print(f"Uterus papers - Mean: {mean_year_uterus:.2f}, Median: {median_year_uterus}, Std Dev: {std_dev_year_uterus:.2f}, Min: {min_year_uterus}, Max: {max_year_uterus}")
```
